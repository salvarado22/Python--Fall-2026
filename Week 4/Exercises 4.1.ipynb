```python
#Exercise 3
#Build a calculator that NEVER crashes —handles all bad input:
#1. Ask user for two numbers (use input())
#2. Ask for an operator: + -* /
#3. Perform the calculation
#4. Handle ALL these error cases with try/except:
#• User enters text instead of a number (ValueError)
#• User tries to divide by zero (ZeroDivisionError)
#• User enters an unknown operator (your own check)
#5. Print the result with 2 decimal places
#6. Keep asking until user types 'quit'
#This is a while loop + try/except + if/elif combined!
#Wrap in:while True: ... then break on 'quit'
```

```python
try:
    number1 = float(input('Enter a number: '))
    number2 = float(input('Enter a number: '))
except ValueError:
    print('This is not a valid set of numbers. Please try again.')

operator = input("Enter operator (+, -, *, /): ").strip().lower()
if operator not in ['+', '-', '*', '/']:
    print('This is not a valid operation. Please try again.')

try:
    if operator == '+':
        result = number1 + number2
    elif operator == '-':
        result = number1 - number2
    elif operator == '*':
        result = number1 * number2
    elif operator == '/':
            # Division by zero check (Handles ZeroDivisionError)
        result = number1 / number2
    else:
        ValueError("Unknown operator")
    print(f'Result: {result:.2f}')
```
