```python
#The tricky ones- these are PCEP favotrites

# // Floor division - rounds DOWN to nearest integer
print(7 // 2) #3 (not 3.5)
print(-7 // 2) # -4 (rounds DOWN, not toward zero!)
print(7.0 // 2) #3.0 (returns float if either operand is float)
```

```python
# % Modulo - remainder after division 
print( 10 % 3) #1 (10 = 3*3 +1)
print(7 % 7)   #0 (divides evenly)
print(-7 % 3)  #2 (sign follows the DIVISOR in Python)
```

```python
# ** Exponential - higher precedence than unary minus!
print(2 ** 10)  #1024
print(-2 ** 2)  # -4
print((-2)**2)  #4 (parenthese change the result)

#PCEP trap: -2 ** 2 --> -2(2**2) = -4 NOT (-2)**2 = 4
```

```python
#a = 6 -> 0110 in binary
#b = 3 -> 0011 in binary
a = 6; b = 3

print(a & b) #AND: 0110 & 0011 = 0010 =2
print (a | b) #OR: 0110 | 0011 = 0111 =7
print(a ^ b) #XQR: 0110 ^ 0011 = 0101 = 5
print(~a) #NOT: ~0110 = -(a+1) = -7
print(a << 1) #Left Shift: 0110 --> 1100 = 12
print(a >> 1) #Right Shift: 0110 --> 0011 = 3
```

```python
# is / is not - identity (same object in memory)
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print (a == b)  #True - same VALUES
print(a is b)   #False - different objects in memory
print( a is c)   #True - c points to the SAME object as a

#None check - ALWAYS use 'is', never '=='
result = None
if result is None:  #correct
    print('No result')

#in / not in - Membership (does item exist in sequence?)
fruits = ['apple', 'banana', 'cherry']
print('apple' in fruits) # True
print('mango' not in fruits) # True

text = 'Hello Python'
print('Python' in text) # True- works on strings too
print('python' in text) #False - case-sensitive!
```

```python
#int() - converts to integer

print(int('42'))  #42 - strong of digits --> int
print(int(3.9))  #3 - float --> int (TRUNCATES, does not round)
print(int(True)) #1 -bool --> int:
#int('3.5')      #ValueError! -string with decimal point fails

#float() - converts to float
print(float('3.14')) #3.14
print(float(5)) #5.0
print(float('inf')) #inf - special value!

#str() - converts to string
print(str(42)) #'42'
print(str(3.14)) #'3.14'
print(str(True)) #'True'

#bool() - what evaluates to False?
print(bool(0)) #False - zero is falsy
print(bool('')) #False - empty string is falsy
#print(bool([])) #False - empty lit is falsy
print(bool(None)) # False - none is falsy
print(bool('0')) #True - non-empty string is TRUTHY
```

```python
#Default: print adds a newline at the end
print('Hello') #Hello\n (moves to next line)

#end= parameter: change what prints at the end
print('Hello', end= ' ') #Hello (space, no newline)
print('World')          # -> Hello World (on same line)
print('A', end= '')
print('B', end='')
print('C')

#sep = paramenter: change separator between arguments
print('a', 'b', 'c') #a b c (default space)
print('a', 'b', 'c', sep=' ') # a b c
print('a', 'b', 'c', sep=', ') #a, b, c

#f-string formatting
pi = 3.14159
print(f'{pi:.2f}') #3.14 (2 decial places)
print(f'{42:5d}') #'    42' (right- align in width 5)
print(f'{42:<5d}') #'42     ' (left-align in width 5)
```

```python
#PCEP Block 1

#Trap 1: -2 ** 2
print(-2 ** 2) # -4 (not 4) ** before unary -

#Trap 2: integer division sign
print(-7 // 2) # -4 (rounds DOWN, not toward zero)

#Trap 3: modulo with negative
print( -7 % 3) #2 (sign follows divisor in Python)

#Trap 4: bool arithmetic
print(True + True) #2 (True == 1 in arithmetic)
print(True * 5) #5

#Trap 5: integer vs float divison 
print(type(4/2)) #<class 'float'> (always float!)
print(type(4 // 2)) #<class 'int'>

#Trap 6: string * int
print('ha' * 3) #'hahaha' - repetition, not error

#Trap 7: int('3.5') is ValueError- must use float() first
print(int(float('3.5'))) #3 - correct appraoch
```

```python
#PCEP Block 2

#Trap 1: range() stop is ALWAYS excluded
list(range(5))  #[0,1,2,3,4] - not 5!
list(range(1,6)) #[1,2,3,4,5]

#Trap 2: for else- else only runs if break never fired
for i in range(3):
    if i==5: break #never fires
else:
    print('done') #prints - loop finished normally

#Trap 3: -= inside loop - not always +=
x = 0
for i in range(1, 4): x -= i
print(x)

#Trap 4: elif stops at first True
n = 90
if n >= 90: print('A')
elif n >= 80: print('B') #never reached - 90 already matched

#Trap 5: Ternary reads left to right (value THEN condition)
x = 5
result = 'big' if x > 3 else 'small' #'big' (value first!)
```
