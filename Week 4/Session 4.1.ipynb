```python
try:
    #Code that might raise an error
    risky_operation()
except ErrorType:
    #RUNS ONLY if that error occurs
    handle_the_error()
else: 
    #Runs ONLY if NO error occurred
    success_operation()
finally:
    #ALWAYS runs - error or not
    cleanup_operation()
```

```python
#Example 1: Catching a specific error

try: 
    result = 10 / 0
except ZeroDivisionError:
    print('Cannot divide by zero!')
print('Program continues...')
```

```python
#Example 2: Catching ValueError from bad input
try:
    age = int(input('Enter your age:'))
    print(f'You are {age} years old')
except ValueError:
    print('Age must be a whole number!')
    
```

```python
#Example 3: Multiple except blocks
try:
    nums = [1, 2, 3]
    result = nums[10] / 0
except IndexError: 
    print('List index our of range!')
except ZeroDivisionError:
    print('Cannot divide by zero!')
```

```python
#Example 3: Multiple except blocks
try:
    nums = [1, 2, 3]
    result = nums[2] / 1
except IndexError: 
    print('List index our of range!')
except ZeroDivisionError:
    print('Cannot divide by zero!')
```

```python
def safe_divide(a, b):
    try: 
        result = a / b   #might raise ZeroDivision Error
    except ZeroDivisionError:
        print('Error: Cannot divide by zero!')
        result = None
    else: 
        #Only runs if NO exception occured
        print(f'Success: {a} / {b} = {result:.2f}')
    finally: 
        #ALWAYS runs - even after an exception
        print('Division attempt complete')
    return result

safe_divide(10, 2)  #success 10 / 2 = 5.00 + 'complete'
safe_divide(10, 0)  #Error: cannot divide...+'complete'
```
