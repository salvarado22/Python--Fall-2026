```python
raw_data=['JOHN SMITH','alice JOHNSON','BOB jones']
clean_data=[]
for name in raw_data:
    clean=name.strip().title()
    clean_data.append(clean)
print(clean_data)
print(f"Client's names:{clean_data}")

#Alternative-handle multiple internal spaces:
for name in raw_data:
    clean="".join(name.split()).title()
    clean_data.append(clean)
print(clean_data)
```

```python
#variables

email1 = 'user@company.com'
email2 = 'invalid.email'
email3 = ' USER@EMAIL.COM '

email4 = email1.strip().lower() #remove spaces
email5 = email2.strip().lower() #remove spaces
email6 = email3.strip().lower() #remove spaces
print(email4)
print(email5)
print(email6)

if '@' in email4 and '.' in email4:
    print('Valid')
    at = email4.index('@')
    user = email4[:at]
    domain = email4[at+1:]
    print(f'User: {user}')
    print(f'Domain: {domain}')

else:
    print('Invalid')
if '@' in email5 and '.' in email5:
    at = email5.index('@')
    user = email5[:at]
    domain = email5[at+1:]
    print(f'User: {user}')
    print(f'Domain: {domain}')
    print('Valid')
else:
    print('Invalid')

if '@' in email6 and '.' in email6:
    print('Valid')
    at = email6.index('@')
    user = email6[:at]
    domain = email6[at+1:]
    print(f'User: {user}')
    print(f'Domain: {domain}')

else:
    print('Invalid')
```
