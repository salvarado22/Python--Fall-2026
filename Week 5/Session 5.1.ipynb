```python
student = {'name': 'Alice', 'age':20, 'grade': 'A'}

#Method 1: Square breacker - KeyError if key missing
print(student['name'])
print(student['age'])
#print(student['gpa']) #KeyError: 'gpa' - crashes

#Method 2: .get() - returns None (or default) if missing
print(student.get('grade')) #A
print(student.get('gpa')) #None - no crash
print(student.get('gpa', 0.0))  #0.0 -custom default

#Check if key exists before accessing
if 'name' in student:
    print(student['name'])  #safe access

#PCEP trap: student['missing'] raises KeyError
#student.get('missing') returns None -no error
```

```python
#Adding, Updating, and Deleting

student = {'name': 'Alice', 'age': 20}

#ADD a new key-value pair
student['grade'] = 'A'  #{'name': 'Alice', 'age': 20, 'grade': 'A'}

#UPDATE an existing value
student['age'] = 21 #overwrite 20 with 21

#Delete a key- del raises KeyError if missing
del student['grade'] #removes key entirely

#.pop() - removes AND returns the value
age = student.pop('age') # age =21 key removed from dict
missing = student.pop('gpa', 0) #safe pop - returns 0 if missing

#.update() - merge another dict in 
extras = {'gpa': 3.9, 'year': 2}
student.update(extras) #adds/overwrites with extras

print(student) #added myself to test out

#Clear all items
student.clear()  #{} -empty dict, not deleted

print(student) #added myself to test out
```

```python
d = {'a': 1, 'b': 2, 'c': 3}

# .keys() -returns all keys as a view
print(d.keys()) #dict_keys(['a', 'b'. 'c']

#.values() -returns all values as a view
print(d.values())  #dict_values([1, 2, 3])

#.items() - returns key-value pairs as tuples
print(d.items()) #dict_items([('a', 1), ('b', 2), ('c', 3)])

#Check membership - 'in' checks KEYS by default
print('a' in d) #True
print(1 in d) #False - 1 is a VALUE not a key
print(1 in d.values()) #True

#len() - number of key-value pairs
print(len(d)) #3
```

```python
#Iterating Over Dictionaries

grades = {'Alice': 95, 'Bob': 87, 'Carol': 92}

#Iterate keys (default - same as 'for k in grades.keys()')
for student in grades:
    print(student) # Alice Bob Carol

#Iterate values only
for score in grades.values():
    print(score)  #95 87 92

#Iterate BOTH key and value - .items() is the PCEP favourite
for student, score in grades.items():
    print(f'{student}: {score}')

#Alice: 95
#Bob: 87
#Carol: 92

#Conditional inside dict iteration
for student, score in grades.items():
    if score >= 90:
        print(f'{student} earned an A')
```

```python
#Common Dict Patterns

#Pattern 1: Frequency counter
words = ['cat', 'dog', 'cat', 'bird', 'dog', 'cat']
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) +1
#'cat': 3, 'dog': 2, 'bird': 1

#Pattern 2: Group by category
students = [('Alice', 'A'), ('Bob', 'B'), ('Carol', 'A'), ('Dave', 'B')]
by_grade = {}
for name, grade in students:
    if grade not in by_grade:
        by_grade[grade] = []
    by_grade[grade].append(name)
#{'A': ['Alice', 'Carol'], 'B': ['Bob', 'Dave']}

#Pattern 3: invert a dict (sqap keys and values)
original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}
#{1: 'a', 2: 'b', 3: 'c']
```
