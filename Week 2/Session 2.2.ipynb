```python
#Without a loop
#Processing 5 sensor readings manually

def process (reading):  #from next week session
    print(reading)
    
readings = [23.1, 24.5, 22.8, 21.9, 24.1]

process(readings[0])
process(readings[1])
process(readings[2])
#... 97 more lines for 100 readings!
```

```python
#The Problem Loops Solve

#With a loop
def process (reading):  #from next week session
    print(reading)
    
readings = [23.1, 24.5, 22.8, 23.4, 25.5]

for reading in readings:
    process(reading)

#same 2 lines for 100 or 1,000,000
```

```python
daily_sales = [120, 85, 200, 95, 175, 60, 230]

#Pattern 1 - Accumulator (build a running total)
revenue = 0
for sale in daily_sales:
    revenue += sale
print(f'Weekly revenue: ${revenue}')
    
```

```python
#Pattern 2- Counter (count days above target)

good_days = 0 # initializing the counter to 0
for sale in daily_sales:
    if sale >= 150:
        good_days += 1
print(f'Days above target: {good_days}')
```

```python
#Pattern 3- Collector (build filtered list)
slow_days = []
for sale in daily_sales:
    if sale in daily_sales:
        if sale < 100:
            slow_days.append(sale)
print(f'Slow days: {slow_days}')
```

```python
#3 Filter - only pass scores (>=60)
scores = [85, 42, 91, 55, 73]
passing = [s for s in scores if s>=60]
print(passing)
```

```python
#4. Tranform AND filter - douvle scores that pass
doubled_pass = [ s * 2 for s in scores if s >= 60]
print(doubled_pass)
```

```python
#5. Conditional expression (ternary) - label each score
labels = ['Pass' if s >= 60 else 'Fail' for s in scores]
print(labels)
```
