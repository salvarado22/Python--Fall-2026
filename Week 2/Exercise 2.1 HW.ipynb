```python
#Exercise 1

grades = [] #empty list
grades.append(95) #adding one number at a time
grades.append(87) #adding one number at a time
grades.append(92) #adding one number at a time
grades.append(78) #adding one number at a time
grades.append(88) #adding one number at a time

print(grades) #new list

count = len(grades) #length of items in list
total = sum(grades)
highest = max(grades)
lowest = min(grades)
average = total / count 

print(f' Number of grades: {count}, Highest grade: {highest}, Lowest grade: {lowest}, Average grade: {average:.1f}')

grades.sort() #sort from lowest to highest
print(grades) #print the new list from the original empty list
```

```python
#Exercise 2

items = ['apples', 'bread', 'milk']

items.append('eggs') #adding an item
items.append('cheese') #adding an item
items.remove('bread') #removing an item
items.insert(0, 'butter') #placing an item in the zero position

print(items)
print(f' Total: {len(items)}')
print(f' First: {items[0]}')
print(f' Last: {items[-1]}')
```

```python
#Exercise 3 Temperature Analyzer

temps = [72, 68, 75, 70, 73, 69, 71] #list

print(temps)
average = sum(temps) / len(temps)  #total / number of items
count1 = len([x for x in temps if x > average])
count2 = len([x for x in temps if x < average])

print(f' Average: {sum(temps) / len(temps):.1f}') #total / number of items
print(f' Highest: {max(temps)}') #max
print(f' Lowest: {min(temps)}')  #min
print(f' Range: {max(temps)-min(temps)}') #highest - lowest
print(f' Days above average: {count1}') 
print(f' Days below average: {count2}')

```
