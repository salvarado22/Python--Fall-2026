```python
#Build a multiplication table with nested loops
for row in range(1, 6):
    for col in range(1, 6):
        product = row * col
        print(f'{product:3d}', end= ' ')
    print()
```

```python
#Student grades - 2D list: [student][subject]
grades = [
    [95, 87, 92], #Alice: Math, English, Science
    [88, 90, 85], #Bob
    [72, 68, 79], #Carol
]
names = ['Alice', 'Bob', 'Carol']

for student_idx, student_grades in enumerate(grades):
    total = 0
    for grade in student_grades: #inner loop: each subkect
        total += grade           #accumulate (pattern from 2.2!)
    average = total / len(student_grades)
    status = 'Pass' if average >= 70 else 'Fail' #conditional from 3.1!
    print(f'{names[student_idx]}: avg={average:.1f} {status}')
```

```python
#Generate combinations - only include valid product sizes

colors = ['red', 'blue', 'green']
sizes = ['small', 'medium', 'large', 'XL']
products = []

for color in colors:                #outer loop
    for size in sizes:              #inner loop
        if size != 'XL' or color == 'blue':   #conditional!
            products.append(f'{size}{color}')
#Only XL blue is allowed - all other XL excluded
#Total combinations without filter: 3 x 4 = 12
#With filter: 3x3 + 1 =10 products
print(f'Products available: {len(products)}')
```

```python
#Complete data processing pipeline -everything combined
raw_data = [85, -1, 92, 150, 73, 88, 45, 91, 200, 60]

#Step 1: Validate data (loop + conditional)
valid = []
for score in raw_data:
    if 0 <= score <= 100:    #only valid scores
        valid.append(score)

#Step 2: Classify each valid score (loop + elif)
for score in valid:
    if score >= 90: grade = 'A'
    elif score >= 80: grade = 'B'
    elif score >= 70: grade = 'C'
    elif score >= 60: grade = 'D'
    else:             grade = 'F'
    print(f' {score}: {grade}')

#Step 3: Statistics (accumulator pattern)
avg = sum(valid) / len(valid)
print(f'Class average: {avg:.1f}  | Valid scores: {len(valid)}/{len(raw_data)}')
```
