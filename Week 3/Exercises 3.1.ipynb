```python
#Exercise 1: Grade Classifier
scores = [95, 87, 42, 73, 68, 91, 55, 80]

for score in scores: 
    if score >= 90:
        grade   = 'A'
        message = 'Excellent!'
    elif score >= 80:
        grade   = 'B'
        message = 'Great work!'
    elif score >= 70:
        grade   = 'C'
        message = 'Good job'
    elif score >= 60:
        grade   = 'D'
        message = 'Needs improvement'
    else:
        grade   = 'F'
        message = 'Please retake'
    print(f"{score} -> {grade} — {message}")
```

```python
#Exercise 2 Eligibility Checker
people = [
    {'name': 'Alice', 'age': 20, 'citizen': True},
    {'name': 'Bob', 'age': 16, 'citizen': True},
    {'name': 'Carol', 'age': 25, 'citizen': False},
    {'name': 'Dave', 'age': 19, 'citizen': True}
]

eligible_count = 0   # Counter for bonus requirement

for person in people: 
    name    = person['name']      # Access dictionary values by key
    age     = person['age']
    citizen = person['citizen']

    if age >= 18 and citizen:
        # AND: both conditions must be True
        print(f"{name}: Eligible to vote")
        eligible_count += 1
    elif age < 18 and citizen:
        # Old enough? No. Citizen? Yes. → age is the only issue
        print(f"{name}: Not eligible — must be 18 or older")
    elif age >= 18 and not citizen:
        # Old enough? Yes. Citizen? No. → citizenship is the only issue
        print(f"{name}: Not eligible — must be a citizen")
    else:
        # Both conditions fail
        print(f"{name}: Not eligible — must be 18 and a citizen")

print(f"\nTotal eligible voters: {eligible_count}")
```

```python
#Exercise 3: FizzBuzz Classic

for n in range(1, 31):
    if n % 3 ==0 and n % 5 ==0:
        print('FizzBuzz', end=' ')
    elif n % 3 == 0:
        print('Fizz', end=' ')
    elif n % 5 ==0:
        print('Buzz', end=' ')
    else:
        print(n, end=' ')
```

```python
#Exercise 4: PCEP Exam

# ── Drill 1: Nested conditionals ─────────────────
x = 8
if x > 5:                  # 8 > 5 True — enter outer if
    if x > 10:             # 8 > 10 False — skip
        print('big')
    else:
        print('medium')    # ← runs: inner else
else:
    print('small')
# OUTPUT: medium

# ── Drill 2: Boolean and/or ───────────────────────
a = True
b = False
if a and b:                # True AND False = False — skip
    print('both')
elif a or b:               # True OR False = True — runs
    print('one')           # ← runs
else:
    print('none')
# OUTPUT: one

# ── Drill 3: Ternary in print ─────────────────────
n = 15
print('fizz' if n % 3 == 0 else n,    # 15%3=0 True → 'fizz'
      'buzz' if n % 5 == 0 else '')   # 15%5=0 True → 'buzz'
# OUTPUT: fizz buzz


```
